<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Customer в C#, который будет представлять информацию о клиентах или покупателях. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Customer:

• Атрибуты: Идентификатор клиента (CustomerId), Имя (Name), Электронная почта (Email).

• Методы:

GetFullName(): метод для получения полного имени клиента. UpdateEmail(string newEmail): метод для обновления электронной почты клиента. ViewProfile(): метод для просмотра профиля клиента. Требования к производным классам:

VIPКлиент (VipCustomer): Должен содержать дополнительные атрибуты, такие как Баланс лояльности (LoyaltyPoints). Метод ViewProfile() должен быть переопределен для отображения дополнительной информации о VIPклиенте. ОбычныйКлиент (RegularCustomer): Должен содержать дополнительные атрибуты, такие как Дата регистрации (RegistrationDate). Метод UpdateEmail() должен быть переопределен для добавления информации о дате последнего обновления электронной почты. ГрупповойКлиент (GroupCustomer) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Название группы (GroupName). Метод GetFullName() должен быть переопределен для отображения названия группы вместо имени клиента.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
public class Customer
{
    public int CustomerId;
    public string Name;

    public Customer(int customerId, string name)
    {
        CustomerId = customerId;
        Name = name;
    }
}

public abstract class BaseCustomer
{
    private int _CustomerId;
    private string _CustomerName;
    private string _Email;

    protected List<Customer> customers;

    public int CustomerId
    {
        get { return _CustomerId; }
        set { _CustomerId = value; }
    }

    public string CustomerName
    {
        get { return _CustomerName; }
        set { _CustomerName = value; }
    }

    public string Email
    {
        get { return _Email; }
        set { _Email = value; }
    }

    public BaseCustomer(int customerId, string customerName, string email)
    {
        CustomerId = customerId;
        CustomerName = customerName;
        Email = email;
        customers = new List<Customer>();
    }
}

public class CustomerManager : BaseCustomer
{
    public DateTime LastUpdated { get; set; }

    public CustomerManager(int customerId, string customerName, string email)
        : base(customerId, customerName, email)
    {
    }

    public virtual void DisplayCustomerInfo()
    {
        Console.WriteLine(
            $"Клиент: {CustomerName}, ID: {CustomerId}, Email: {Email}");
    }

    public virtual void AddCustomer(Customer customer)
    {
        customers.Add(customer);
        LastUpdated = DateTime.Now;

        Console.WriteLine(
            $"Клиент '{customer.Name}' добавлен.");

        Console.WriteLine(
            $"Список клиентов обновлен: {LastUpdated}");
    }

    public virtual void RemoveCustomer(Customer customer)
    {
        if (customers.Remove(customer))
        {
            Console.WriteLine(
                $"Клиент '{customer.Name}' удален.");
        }
        else
        {
            Console.WriteLine(
                $"Клиент '{customer.Name}' не найден.");
        }

        LastUpdated = DateTime.Now;
    }
}

public interface IManageable
{
    void ChangeCustomer(string newName);
}

public class VipCustomer : CustomerManager, IManageable
{
    private string _PhoneNumber;
    private int _LoyaltyPoints;

    public string PhoneNumber
    {
        get { return _PhoneNumber; }
        set { _PhoneNumber = value; }
    }

    public int LoyaltyPoints
    {
        get { return _LoyaltyPoints; }
        set { _LoyaltyPoints = value; }
    }

    public VipCustomer(
        int customerId,
        string customerName,
        string email,
        string phoneNumber)
        : base(customerId, customerName, email)
    {
        PhoneNumber = phoneNumber;
        LoyaltyPoints = 0;
    }

    public override void DisplayCustomerInfo()
    {
        Console.WriteLine(
            $"VIP-клиент: {CustomerName}, ID: {CustomerId}, " +
            $"Email: {Email}, Телефон: {PhoneNumber}, " +
            $"Бонусные баллы: {LoyaltyPoints}");
    }

    public void ChangeCustomer(string newName)
    {
        Console.WriteLine(
            $"VIP-клиент сменил имя с '{CustomerName}' на '{newName}'.");

        CustomerName = newName;
    }

    public void AddLoyaltyPoints(int points)
    {
        LoyaltyPoints += points;

        Console.WriteLine(
            $"VIP-клиенту начислено {points} бонусных баллов.");
    }

    public void AddLoyaltyPoints(int points, string reason)
    {
        LoyaltyPoints += points;

        Console.WriteLine(
            $"VIP-клиенту начислено {points} баллов. Причина: {reason}.");
    }

    public void GetVipBenefits()
    {
        Console.WriteLine(
            $"VIP-клиент {CustomerName} получает специальные предложения.");
    }
}

public class RegularCustomer : CustomerManager, IManageable
{
    private DateTime _RegistrationDate;
    private int _PurchaseCount;

    public DateTime RegistrationDate
    {
        get { return _RegistrationDate; }
        set { _RegistrationDate = value; }
    }

    public int PurchaseCount
    {
        get { return _PurchaseCount; }
        set { _PurchaseCount = value; }
    }

    public RegularCustomer(int customerId,
        string customerName,
        string email)
        : base(customerId, customerName, email)
    {
        RegistrationDate = DateTime.Now;
        PurchaseCount = 0;
    }

    public override void DisplayCustomerInfo()
    {
        Console.WriteLine(
            $"Обычный клиент: {CustomerName}, ID: {CustomerId}, " +
            $"Email: {Email}, Покупок: {PurchaseCount}, " +
            $"Дата регистрации: {RegistrationDate}");
    }

    public void ChangeCustomer(string newName)
    {
        Console.WriteLine(
            $"Имя клиента '{CustomerName}' изменено на '{newName}'.");

        CustomerName = newName;
    }

    public void MakePurchase()
    {
        PurchaseCount++;

        Console.WriteLine(
            $"Клиент {CustomerName} совершил покупку.");
    }

    public void MakePurchase(string productName)
    {
        PurchaseCount++;

        Console.WriteLine(
            $"Клиент {CustomerName} купил товар '{productName}'.");
    }

    public void DisplayRegistrationInfo()
    {
        Console.WriteLine(
            $"Клиент {CustomerName} зарегистрирован: {RegistrationDate}");
    }
}

public class GroupCustomer : CustomerManager
{
    private string _GroupName;
    private int _GroupSize;

    public string GroupName
    {
        get { return _GroupName; }
        set { _GroupName = value; }
    }

    public int GroupSize
    {
        get { return _GroupSize; }
        set { _GroupSize = value; }
    }

    public GroupCustomer(
        int customerId,
        string customerName,
        string email,
        string groupName)
        : base(customerId, customerName, email)
    {
        GroupName = groupName;
        GroupSize = 0;
    }

    public override void DisplayCustomerInfo()
    {
        Console.WriteLine(
            $"Групповой клиент: {CustomerName}, ID: {CustomerId}, " +
            $"Email: {Email}, Группа: {GroupName}, " +
            $"Участников: {GroupSize}");
    }

    public override void AddCustomer(Customer customer)
    {
        base.AddCustomer(customer);
        GroupSize++;

        Console.WriteLine(
            $"В группу '{GroupName}' добавлен участник. " +
            $"Количество участников: {GroupSize}");
    }

    public void AddGroupMember(Customer customer)
    {
        AddCustomer(customer);
    }

    public void DisplayGroupInfo()
    {
        Console.WriteLine(
            $"Группа: {GroupName}, количество участников: {GroupSize}");
    }

    public void ChangeGroupName(string newGroupName)
    {
        Console.WriteLine(
            $"Название группы изменено с '{GroupName}' на '{newGroupName}'.");

        GroupName = newGroupName;
    }
}

// Создание объектов

CustomerManager myCustomerManager =
    new CustomerManager(1, "Анна Иванова", "anna@mail.com");

VipCustomer myVipCustomer =
    new VipCustomer(2, "Алексей Петров", "alex@mail.com", "+79990000001");

RegularCustomer myRegularCustomer =
    new RegularCustomer(3, "Мария Смирнова", "maria@mail.com");

GroupCustomer myGroupCustomer =
    new GroupCustomer(4, "Иван Сидоров", "ivan@mail.com", "Постоянные клиенты");

// Работа с VIP-клиентом

myVipCustomer.DisplayCustomerInfo();
myVipCustomer.AddLoyaltyPoints(100);
myVipCustomer.AddLoyaltyPoints(50, "Покупка на день рождения");
myVipCustomer.GetVipBenefits();
myVipCustomer.ChangeCustomer("Алексей Петрович");

Console.WriteLine();

// Работа с обычным клиентом

myRegularCustomer.DisplayCustomerInfo();
myRegularCustomer.MakePurchase();
myRegularCustomer.MakePurchase("Ноутбук");
myRegularCustomer.DisplayRegistrationInfo();

Console.WriteLine();

// Работа с групповым клиентом

myGroupCustomer.DisplayCustomerInfo();

myGroupCustomer.AddGroupMember(
    new Customer(5, "Елена Кузнецова"));

myGroupCustomer.AddGroupMember(
    new Customer(6, "Дмитрий Волков"));

myGroupCustomer.DisplayGroupInfo();
myGroupCustomer.ChangeGroupName("VIP-группа");

Console.WriteLine();

// Полиморфизм

CustomerManager[] customerList =
{
    myCustomerManager,
    myVipCustomer,
    myRegularCustomer,
    myGroupCustomer
};

foreach (CustomerManager customer in customerList)
{
    customer.DisplayCustomerInfo();
    Console.WriteLine();
}
```
